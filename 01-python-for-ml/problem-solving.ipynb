### digit summation problem


```python
numbers=input()
print(numbers)
split_numbers=numbers.split()
print(split_numbers)
number1=int(split_numbers[0])
number2=int(split_numbers[1])
number1stReminder=number1%10
print(number1stReminder)
number2ndReminder=number2%10
print(number2ndReminder)
sum_of_reminders=number1stReminder+number2ndReminder
print(sum_of_reminders)
```
