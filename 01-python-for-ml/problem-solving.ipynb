### C. Even, Odd, Positive and Negative
## Given N numbers. Count how many of these values are even, odd, positive and negative

```python
given_number_value=int(input())
given_number_for_even_odd_neg_pos=input()
split_given_number_for_even_odd_neg_pos=given_number_for_even_odd_neg_pos.split()
negative=0
positive=0
even=0
odd=0
for i in range(given_number_value):
    number=int(split_given_number_for_even_odd_neg_pos[i])
    if(number%2==0 or number%4==0):
        even+=1
    if(number<0):
       negative+=1
    if(number>0):
        positive+=1
    if(number%2!=0 and number%4!=0):
        odd+=1
print("Even:",even)
print("Odd:",odd)
print("Positive:",positive)
print("Negative:",negative)
```
