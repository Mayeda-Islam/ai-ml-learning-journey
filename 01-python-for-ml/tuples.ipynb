```python
# declaration
tuple_of_numbers=(1,2,3,4,5)
tuple_of_strings=("apple","banana","cherry")
tuple_of_mixed=(1,"apple",2,"banana",3,"cherry")
print(tuple_of_numbers)
print(tuple_of_strings)
print(tuple_of_mixed)
print(type(tuple_of_numbers))
print(type(tuple_of_strings))
print(type(tuple_of_mixed))
```

```python
# conversion from list to tuple
list_of_numbers=["1",2,3,False,5]
tuple_from_list=tuple(list_of_numbers)
print(tuple_from_list)
print(type(tuple_from_list))
print(type(tuple_from_list[0]))  # 1
print(tuple_from_list[0:3])  # 1
```

```python
# immutable
tuple=(5,7,9,0,"str",False)
print(tuple)
# tuple[0]=1 ,cannot assign to tuple, because tuple is immutable
# print(tuple)
# tuple.append(1)  # cannot append
print(tuple.count(5))  # 1
print(tuple.index("str"))  # 4
```
