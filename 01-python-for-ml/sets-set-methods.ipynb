### Set : In built data structure
## A set is a collection of unique items in Python.
## Unordered: items do not stay in a fixed order
## Unique: duplicates are automatically removed
## Mutable: you can add or remove items
## No indexing: you cannot access items by position like s[0]


```python
s = {1, 2, 3, 4}
print(s)   # {1, 2, 3, 4}
print(type(s))   # <class 'set'>
```

```python
tuple=()
print(type(tuple))   # <class 'tuple'>
list=[]
print(type(list))   # <class 'list'>
C={}
print(type(C))   # <class 'dict'>
S= set()
print(type(S))   # <class 'set'>   # <class 'set'>
```

```python
set1 = {1, 2, 3,2,3,4, 4}
print(set1)   # {1, 2, 3, 4}  # duplicate values are removed
if 2 in set1:   
    print("2 is present in the set")   # 2 is present in the set
else:
    print("2 is not present in the set")   # 2 is not present in the set
    
sum=0
for i in set1:
    sum+=i
print(sum)   # 10
```

## Set Method

```python
#mutable
set2={1, 2, 3, 4}
set2.add(5)
set2.add(7)
print(set2)   # {1, 2, 3, 4, 5, 7}

# sort 
set3={7,9,4,7,8,2,0,1}
print(set3)   # {0, 1, 2, 4, 7, 8, 9}

# pop
set2.pop()
print(set2)   # {2, 3, 4, 5,

# remove,targeted element remove
set2.remove(3)
print(set2)   # {2, 4, 5, 7}

# 
my_set = {"apple", "banana", "cherry"}

# pop() 
removed_element = my_set.pop()

print("বাদ যাওয়া উপাদান:", removed_element)
print("বাকি থাকা সেট:", my_set)
```

### union

```python
a={1, 2, 3, 4}
b={3,1,2, 4, 5, 6}
# union
print(a.union(b))
# intersection
print("intersection:",a.intersection(b))
# disjoint
print("disjoint:",a.isdisjoint(b))  #disjoint means no common elements, so it will return False
# subset
print("subset:",a.issubset(b))  # subset means all elements of a are in b, so it will return True
# superset
print("superset:",b.issuperset(a))  # superset means all elements of a are in a
```
