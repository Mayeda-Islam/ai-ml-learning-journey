### class and object
## 

```python
class Phone:
    categories="Electronics"
    # constructor
    def __init__(self,name,price,battery_percentage=100):
        self.name=name
        self.price=price
        self.battery_percentage=battery_percentage
    # methods
    def charge(self, hour):
        self.battery_percentage += hour
        print("battery now:", self.battery_percentage)
    def chapters(self,total_photo):
        if(self.battery_percentage<=0):
            print("no battery")
           
        else:
            self.battery_percentage-=total_photo
            print("battery left",self.battery_percentage)    
        
apple=Phone("iphone",10000,90)
# class attribute
apple.categories="Electronics different"
# instance attribute
motorolla=Phone("fusionq",8000,80)
print(apple.name)
print(motorolla.categories)
motorolla.charge(10)
```
