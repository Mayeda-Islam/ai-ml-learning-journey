Comparison operators

```python
# == → equal to
# != → not equal to
# > → greater than
# < → less than
# >= → greater than or equal to
# <= → less than or equal to
a=10;b = 5

print(a == b)    # False
print(a != b)    # True
print(a > b)     # True
print(a < b)     # False
print(a >= 10)   # True
print(b <= 4)    # False
```

Logical operators

- and: returns True if both conditions are True
- or: returns True if at least one condition is True
- not: reverses the result

```python
# Logical operators
# and, or, not

x = 10
y = 20
z = 30

print(x < y and y < z)   # True
print(x > y and y < z)    # False
print(x < y or y > z)     # True
print(not(x < y))         # False
```

```python
Operator Precedence and Associativity
1) Precedence
Precedence means which operator runs first.
2) Associativity
Associativity means whether the operator is left-associative or right-associative.
Precedence = priority
Associativity = order when same priority
```

```python
print(10 + 5 * 2)
# * has higher precedence than +
# So calculation is:
# 5 * 2 = 10
# 10 + 10 = 20

print((10 + 5) * 2)

# Precedence order: *, /, //, % first; then +, -
print(10 / 5 + 3 * 2 - 1)   # (10/5) + (3*2) - 1 = 2 + 6 - 1 = 7.0
print(10 // 5 + 3 * 2 - 1)  # (10//5) + (3*2) - 1 = 2 + 6 - 1 = 7
```

```python
# Associativity means if operators have the same precedence, which one is evaluated first.

print(10 - 5 - 2)
print(10 - (5 - 2)) #Because - is left-associative:
print(8 / 4 / 2) 
print(8 / (4 / 2)) #Because / is left-associative:
print(2 ** 3 ** 2)
print((2 ** 3) ** 2) #Because ** is right-associative:
```
