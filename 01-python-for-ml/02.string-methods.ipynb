## 1) Indexing
### Indexing means getting one character from a string by its position.

```python
text = "Python"

print(text[0])   # P
print(text[1])   # y
print(text[-1])  # n
print(text[-2])  # o

# Index starts from 0
# Negative indexes count from the end:
# -1 = last character
# -2 = second last
```

## 2) Slicing
### Slicing means getting a part of the string

```python
text = "Python"

print(text[0:3])   # Pyt
print(text[2:6])   # thon
print(text[:4])    # Pyth
print(text[2:])    # thon
print(text[:])     # Python
print(text[::2])   # Pto
print(text[::-1])   # nohtyP


# text[start:end] → from start up to, but not including, end
# text[:4] → from beginning to index 4
# text[2:] → from index 2 to end
# text[::2] → step by 2
```

```python
sentence = "I love Python"

print(sentence[0])      # I
print(sentence[2:6])    # love
print(sentence[-6:])    # Python
print(sentence[::2])    # Ilv yhn
```

### String Methods

```python
text = "Welcome to Python learning"
text1 = "Welcome to Python learning python"
processed_text=text.lower()  # welcome to python learning
processed_text1=text1.lower()  
capitalized_text=text.capitalize()  # Welcome to python learning

print(len(text))  # 22
print('python' in text)  # False
print('python' in processed_text)  # True
print('Python' in text)  # True

# substring index
print(text.find('Python'))  # 13
# substring last index,meaning the last occurrence of the substring
print(text1.rfind('python'))  # 13
# count of substring
print(processed_text1.count('python'))  # 2

# replace substring,string are immutable, so it will return a new string
new_text = text.replace('Python', 'ML')  # Welcome to Java learning 000
print(new_text)
```

### Splitting,joinning,formatting

```python
prompt = "Enter your name: "
token=prompt.split()  # ['Enter', 'your', 'name:']
print(token)  
print(type(token))  # <class 'list'>

# join
joined_string = '-'.join(token)  # Enter your name:
print(joined_string)  # Enter your name:
print(type(joined_string))  # <class 'str'>
```
