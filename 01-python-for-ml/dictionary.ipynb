### Dictionary:A dictionary is a collection of key-value pairs
## Keys are unique
## Values can be anything
## Dictionaries are mutable
## They are unordered in older versions, but Python 3.7+ keeps insertion order
## Access value using key

```python
## declation
dict1={}
print(type(dict1))   # <class 'dict'>
dict2={"name":"John", "age":30, "city":"New York","address":[ 10,20,30]}
print(dict2,type(dict2))   # <class 'dict'>
```

```python
# accessing the key
print(dict2["name"])
print(dict2["age"])
print(dict2["city"])
print(dict2.get("name"))
print(dict2.get("age"))
print(dict2.get("city"))

# update the value of a key
dict2["name"]="Jane"
print(dict2)
dict3={"name":"John", "age":30, "city":"New York","address":[ 10,20,30],"age":40}
print(dict3)   # {'name': 'John', 'age': 40, 'city': 'New York', 'address': [10, 20, 30]},bcz age key is already present, so it will update the value of age key to 40

# accessing the value by get function
print(dict3.get("name"))   # John
print(dict3.get("age",35))   # add default value if key is not present, so it will return 40
```

```python
#adding new key-value pair
dict4={"name":"John", "age":30, "city":"New York","address":[ 10,20,30],"email":"john@example.com"}
print(dict4)   # {'name': 'John', 'age': 30, 'city': 'New York', 'address': [10, 20, 30], 'email': '
dict4["phone"]="123-456-7890"
print(dict4)  
dict4.update({"weight":60,})
print(dict4)  
```

```python
# deleting a key-value pair
del dict4["email"]
print(dict4)
```

```python
# copying a dictionary into another
new_dict=dict4.copy()
print(new_dict)
```

### Iteration over dictionary

```python
key_list=dict4.keys()
print(key_list)   # dict_keys(['name', 'age', 'city', 'address
dict4["gender"]="male"
print(key_list)   # dict_keys(['name', 'age', 'city', 'address
```

```python
# keys() method returns a view object that displays a list of all the keys in the dictionary. The view object is dynamic, meaning that it reflects any changes made to the dictionary. In this case, when we add a new key-value pair to dict4, the view object key_list is updated to include the new key 'gender'.
items=dict4.items()
print(items)  # dict_items([('name', 'John'), ('age', 30), ('city', 'New York'), ('address', [10, 20, 30]), ('phone', '123-456-7890'), ('weight', 60), ('gender', 'male')])
```

```python
# iteration
for key,value in dict4.items():
    print(key,value)
```

### Dictionary Comprehensive

```python
print(list(range(10)))
square={x:x**2 for x in range(10)}
#meaning of this line is to create a dictionary where the keys are numbers from 0 to 9 
# and the values are their squares. The dictionary comprehension iterates over
# each number x in the range of 10 
# and assigns x as the key and x**2 as the value.
print(square)
```
