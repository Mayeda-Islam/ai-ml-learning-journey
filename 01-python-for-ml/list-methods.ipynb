```python
#declaration
list_of_numbers=[1,2,3,4,5]
list_of_strings=["apple","banana","cherry"]
list_of_mixed=["apple",1,"banana",2,"cherry",3]
print(list_of_numbers)
print(list_of_strings)
print(list_of_mixed)
print(type(list_of_numbers))
print(type(list_of_strings))
print(type(list_of_mixed))
print(type(list_of_mixed[0]))
print(type(list_of_mixed[1]))

# 2d list
list_2d=[[1,2,3],[4,5,6],[7,8,9]]
print(list_2d)
print(list_2d[0])
print(list_2d[1][2])
```

```python
list=[1, 2,2, 3, 4, 5]
print(list)
print(type(list))
print(list[0])  # 1
print(list[0:2])  # [1, 2]

# adding an element to the list
list.append(6)
print(list)

# removing an element from the list 
list.remove(2)
print(list)

# inserting an element at a specific index
list.insert(2, 10)
print(list)

# deleting an element at a specific index
print(list.pop())

# sort the list in ascending order
list.sort()
print(list)

# sort the list in descending order
list.sort(reverse=True)
print(list)

```
